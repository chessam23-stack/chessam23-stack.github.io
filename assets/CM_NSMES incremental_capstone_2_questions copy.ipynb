```python
#We want to write a function that will return average age by gender
#check iff somethign is na to not na

#If something is not an na value, if there is an age, return the age.
#else--the age is na.....
    #if the gender value is male return the avg_age_male value
    #else-meaning that the gender value is not male, aka female, then return avg_age_female value

#This will use lambda



def fill_avg_age(gender_val, age_val):
    if not pd.isna(age_val):
        return age_val
    else:
        if gender_val == 'male':
            return avg_age_male
        else:
            return avg_age_female
```

```python
#We calling apply off the entire data frame
#in lambda x - the entire data frame med_df will look for gender or age
#then we have to give it an axis,, we want it to move down columns = 1, rows = 0

med_df['age_by_gender'] = med_df.apply(lambda x:fill_avg_age(x['gender'], x['age']), axis = 1)
```
