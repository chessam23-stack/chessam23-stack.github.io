```python
x = [1,5,2,4,3,2]  #-> 2.5
y =  [1,2,3,4,5,6,7,15] # -> 4.5
z = [8,4,16,14,20,1,34] #-> 14
w = [1,3,2,4,3,5,4,6,5,7,6] #-> 4
```

```python
  # 1.sort the list, lowest to highest
    # use lst.sort()
    #use floor division--divides the list into two parts, so for second_mid_length it divides it in 2 parts, and then 
    # Odd number of values
    #Even number of values
    #Find the middle two values, and take their average -> 9


def find_median(lst):
    lst.sort()

    mid_index = len(lst) // 2
    second_mid_index = len(lst) //2-1
    
    if len(lst) == 0:
        return "No list."
        
 
    elif len(lst) %2 == 0:
        return (lst[mid_index] + lst[second_mid_index])/2
        
        
    else:
        return lst[mid_index]
 
```
